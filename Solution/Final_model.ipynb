### Simulation und Otpimierung eines Post-Verteilzentrums mit PV-Anlage, Batterie und E-Truck-Flotte
Größe und Art des Speichers hängt von folgenden Bedingungen ab:
- Verbrauchskurve des Standortes auf Basis der Ladefrequenz und -dauer der Fahrzeuge 
- Lichteintrag im Jahresverlauf in Verbindung mit dem Wetter (auf Basis realem Wetter und Wetterforecast) 
- Strompreis (Spotmarkt und Einspeise- wie Bezugstarife) des nächsten Tages, 
- Größe der PV Anlage (optimiert auf Eigenbedarf und nicht Überschusseinspeisung)
- Tägliche km Leistung je Fahrzeug
- Täglicher Energiebedarf je Fahrzeug
- Lastkurve der Anlage

### Optimierungsparameter
- Vermeidung von Netzspitzen
- Nutzung der Preisschwankungen (Spotmarkt mit Laden bei niedrigen oder gar negativen Preisen, etc.)
- Optimierung des Einspeisens des Stromüberschusses
- Einbeziehung der Wetterprognose, um Randbedingungen der Optimierungsfunktion zu tweaken (Netzanteil, Max Ladestrom, Max Preis Netzanteil...)
- Einbeziehung von stationärem Speicher und optimale Bewirtschaftung hinsichtlich Spotmarktpreise o	Kosten der Anlage (PV-Anlage und Speicher) in verschiedenen Dimensionen (Abschreibung, Verzinsung, Instandhaltung, Wartung, Versicherung) versus Einnahmenströme (aus Kostenersparnis von Strombezügen und aktiven Einnahmen aus Einspeisung) o	Netzkosten (Leistungspreis, Kosten mögliche Erweiterung Netzanschluss, …)
